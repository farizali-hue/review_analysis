```python
import os
from dotenv import load_dotenv
from openai import OpenAI

# Load environment variables from .env file
load_dotenv()

# Initialize OpenAI client with API key from env
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

```python
def generate_negative_issue(title, content):
    """
    Generate a short word negative issue from title and content using OpenAI Responses API
    """
    try:
        prompt = f"""Based on the following product review, identify ONE negative issue in a short word only. 
        
Title: {title}
Content: {content}

Respond with only short word describing the negative issue (e.g., 'low durability', 'low quality', 'not so comfort', 'not sizing', etc.)"""
        response = client.responses.create(
            model="gpt-5.2",
            input=prompt,
            max_output_tokens=300,
            temperature=0.3,
        )
        text = response.output_text.strip()
        return text.split()[0].lower()
    except Exception as e:
        print(f"Error: {e}")
        return "unknown"
```

```python
rows = []
for i in range(5):
    # Test with one example first
    sample_row = reviews_df_low_ratings.iloc[i]
    test_issue = generate_negative_issue(sample_row['Title'], sample_row['Content'])
    l = [sample_row['Asin'], sample_row['Title'], sample_row['Content'], test_issue]
    rows.append(l)
```

```python
results_df = pd.DataFrame(rows, columns=['Asin', 'Title', 'Content', 'Negative Issue'])
results_df.to_csv("D:/analysis/negative_issues.csv", index=False)
```

```python
# Apply the function to all rows to create the 'Negative Issue' column
reviews_df_low_ratings['Negative Issue'] = reviews_df_low_ratings.apply(
    lambda row: generate_negative_issue(row['Title'], row['Content']), 
    axis=1
)

# Display the results
print("Negative Issues generated!")
print(reviews_df_low_ratings[['Title', 'Content', 'Negative Issue']].head(10))
```
