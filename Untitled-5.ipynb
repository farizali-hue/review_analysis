```python
def generate_negative_issue(title, content):
    """
    Generate a one-word negative issue from title and content using Gemini API
    """
    try:
        prompt = f"""Based on the following product review, identify ONE negative issue in a single word only. 
        
Title: {title}
Content: {content}

Respond with only one word describing the negative issue (e.g., 'durability', 'quality', 'comfort', 'sizing', etc.)"""
        
        response = openai.ChatCompletion.create(
            model="gpt-4o-mini",
            messages=[{"role": "user", "content": prompt}],
            max_tokens=50
            temperature=0.3)
        text = response.choices[0].message['content'].strip()
        return text.split()[0].lower()  # Get first word and lowercase
    except Exception as e:
        print(f"Error: {e}")
        return "unknown"

# Test with one example first
test_issue = generate_negative_issue(
    reviews_df_low_ratings.iloc[0]['Title'],

    reviews_df_low_ratings.iloc[0]['Content']
    print(f"Test result: {test_issue}"))
```

```python
# Apply the function to all rows to create the 'Negative Issue' column
reviews_df_low_ratings['Negative Issue'] = reviews_df_low_ratings.apply(
    lambda row: generate_negative_issue(row['Title'], row['Content']), 
    axis=1
)

# Display the results
print("Negative Issues generated!")
print(reviews_df_low_ratings[['Title', 'Content', 'Negative Issue']].head(10))
```
